```python
import pandas as pd
from ydata_profiling import ProfileReport

# 1. Charger votre jeu de données CSV
df = pd.read_csv("penguins.csv")

# 2. Générer le rapport d'analyse
profile = ProfileReport(df, title="Rapport d'analyse de données")

# 3. Afficher le rapport directement dans le notebook
profile.to_notebook_iframe()

# 4. Exporter le rapport sous forme de fichier HTML autonome
profile.to_file("rapport_analyse.html")
```
